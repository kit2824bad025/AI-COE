```python

# Transformer-Based Next-Token Prediction
# Alternative Version Using Self-Attention

# 1. Import libraries
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras import layers, Model
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# 2. Sample dataset
sentences = [
    "python is a popular programming language",
    "python is useful for data science",
    "data science uses machine learning",
    "machine learning is a part of artificial intelligence",
    "artificial intelligence can solve complex problems",
    "neural networks are used in deep learning",
    "deep learning uses neural networks",
    "transformers are powerful language models",
    "attention helps transformers understand context",
    "natural language processing works with text",
    "programming helps us build software applications",
    "machine learning discovers patterns in data",
    "deep learning is used in computer vision",
    "attention mechanisms improve language models",
    "python supports artificial intelligence projects"
]

# 3. Tokenization
tokenizer = Tokenizer(oov_token="<UNK>")
tokenizer.fit_on_texts(sentences)

sequences = tokenizer.texts_to_sequences(sentences)
vocab_size = len(tokenizer.word_index) + 1

print("Vocabulary size:", vocab_size)

# 4. Generate next-token training samples
X_data = []
y_data = []

for seq in sequences:
    for i in range(1, len(seq)):
        X_data.append(seq[:i])
        y_data.append(seq[i])

max_length = max(len(seq) for seq in X_data)

X = pad_sequences(
    X_data,
    maxlen=max_length,
    padding="pre",
    truncating="pre"
)
y = np.array(y_data)

print("Training samples:", len(X))
print("Input shape:", X.shape)

# 5. Create Transformer model
embedding_dim = 32
num_heads = 2

inputs = layers.Input(shape=(max_length,))

# Token embeddings
token_embeddings = layers.Embedding(
    vocab_size, embedding_dim, mask_zero=True
)(inputs)

# Positional embeddings
positions = tf.range(start=0, limit=max_length)
position_embeddings = layers.Embedding(
    max_length, embedding_dim
)(positions)

x = token_embeddings + position_embeddings

# Self-attention layer
attention_output = layers.MultiHeadAttention(
    num_heads=num_heads,
    key_dim=16,
    dropout=0.1
)(
    query=x,
    value=x,
    key=x,
    use_causal_mask=True
)

# Residual connection and normalization
x = layers.Add()([x, attention_output])
x = layers.LayerNormalization()(x)

# Feed-forward network
ffn = layers.Dense(64, activation="relu")(x)
ffn = layers.Dense(embedding_dim)(ffn)

x = layers.Add()([x, ffn])
x = layers.LayerNormalization()(x)

# Select the final sequence position
x = layers.Lambda(lambda tensor: tensor[:, -1, :])(x)

# Predict the next token
outputs = layers.Dense(
    vocab_size, activation="softmax"
)(x)

model = Model(inputs, outputs)

# 6. Compile model
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# 7. Train model
history = model.fit(
    X, y,
    epochs=100,
    batch_size=16,
    validation_split=0.1,
    verbose=1
)

print("\nTraining completed!")

# 8. Plot training and validation loss
plt.figure(figsize=(8, 4))
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Transformer Training Performance")
plt.legend()
plt.grid(True)
plt.show()

# 9. Predict the next word
def generate_text(start_text, num_words=5):
    result = start_text

    for _ in range(num_words):
        token_ids = tokenizer.texts_to_sequences([result])[0]

        padded = pad_sequences(
            [token_ids],
            maxlen=max_length,
            padding="pre",
            truncating="pre"
        )

        probabilities = model.predict(padded, verbose=0)[0]

        # Exclude the padding and unknown-token IDs
        probabilities[0] = 0
        unknown_id = tokenizer.word_index.get("<UNK>")
        if unknown_id is not None:
            probabilities[unknown_id] = 0

        next_id = int(np.argmax(probabilities))
        next_word = tokenizer.index_word.get(next_id)

        if next_word is None:
            break

        result += " " + next_word

    return result

# 10. Test predictions
test_inputs = [
    "python is",
    "machine learning",
    "deep learning",
    "artificial intelligence"
]

for sentence in test_inputs:
    print("\nInput:", sentence)
    print("Generated text:", generate_text(sentence, 5))

# 11. Save model
model.save("friend_transformer_model.keras")
print("\nModel saved successfully!")
```
